```python
df = pd.read_csv('places.csv')
```

```python
df["state"] = ""
df["city"] = ""
df["area"] = ""

def appl (in_row):
    arr = in_row['place'].split(";")
    
    if len(arr) == 3:
        in_row['state'] = arr[0] 
        in_row['city'] = arr[1] 
        in_row['area'] = arr[2] 

    elif len(arr) == 2:
        in_row['state'] = arr[0] 
        in_row['city'] = arr[1] 
        in_row['area'] = "" 
    
    else:
        in_row['state'] = arr[0] 
        in_row['city'] = "" 
        in_row['area'] = "" 
    return in_row  

df = df.apply(appl, axis=1)
```
